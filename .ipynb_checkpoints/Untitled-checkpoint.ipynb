```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dat = pd.read_csv('gapper_test_2.csv')
dat
```

```python
tracker_df = []
for symbol, group in dat.groupby("symbol"):
    days = 0
    bestDay = days
    worstDay = days
    bestPriceAfterBuy = group.iloc[0]["pm_last"]
    worstPriceAfterBuy = group.iloc[0]["pm_last"]
    for col,row in group.iterrows():
        if (row["pm_last"] > bestPriceAfterBuy):
            bestDay = days
            bestPriceAfterBuy = row["pm_last"]
        if (row["pm_last"] < worstPriceAfterBuy):
            worstDay = days
            worstPriceAfterBuy = row["pm_last"]
        days += 1
    buyRow = group.iloc[0].copy()
    buyRow['best_price'] = bestPriceAfterBuy
    buyRow['worst_price'] = worstPriceAfterBuy
    buyRow['best_day'] = bestDay
    buyRow['worst_day'] = worstDay
    tracker_df.append(buyRow)
tracked = pd.DataFrame(tracker_df).reset_index(drop=True).drop(["Unnamed: 0"], axis=1)
tracked
```

```python
good = tracked[tracked["best_day"] > 0]
bad = tracked[tracked["best_day"] == 0]
```

```python
good['best_day'].value_counts()
```

## Looks like the first day is the best so far

```python
bad["worst_day"].value_counts()
```

## Gap direction does not appear to be a good indicator. Though, it is slightly favorable towards down gaps. Down gaps slightly more likely to have a positive day. And up gaps slightly more likely to never go positive in our time window

```python
good['gap_up'] = good['pm_gap_pct'] > 0
good['gap_up'].value_counts()
```

```python
bad['gap_up'] = bad['pm_gap_pct'] > 0
bad['gap_up'].value_counts()
```

## Let's combine them both, add a new field to show they went positive and do some modeling. 

## So far, more likely to go positive. But this could just be the aggregate movement of the market for the week, and this doesn't take into consideration how much it went positive. It's possible that we go positive such a little amount that we lose money on the fees processing. 

```python
tracked['posi'] = tracked['best_day'] > 0
tracked['posi'].value_counts()
```

```python
tracked['price_diff'] = tracked['best_price'] - tracked['pm_last']
tracked['has_pm_news'] = tracked['has_pm_news'].astype(int)
tracked['posi'] = tracked['posi'].astype(int)
```

```python
tracked
```

## I don't see any strong correlation between the metrics. Let's just run a linear model and see what there is. 

```python
sns.scatterplot(x=tracked['pm_gap_pct'], y=tracked['price_diff'])
plt.show()
```

```python
import statsmodels.api as sm

# 2. Separate independent variables (X) and dependent variable (y)
X = tracked[["open_price", "pm_gap_pct", "has_pm_news", "best_day", "worst_day"]]
y = tracked["price_diff"]

# 3. Add constant for intercept term (statsmodels does NOT include an intercept by default)
X = sm.add_constant(X)

# 4. Fit the Ordinary Least Squares model
model = sm.OLS(y, X).fit()

# 5. View regression results
print(model.summary())
```

## Looks like wether it has new and what day is the best day are the best indicators. What if we restrict to gappers only and also play with the gap percentage

```python
gappers = tracked[tracked['pm_gap_pct'] > 0.4]
gappers.shape

# 2. Separate independent variables (X) and dependent variable (y)
X = gappers[["open_price", "pm_gap_pct", "has_pm_news", "best_day", "worst_day"]]
y = gappers["price_diff"]

# 3. Add constant for intercept term (statsmodels does NOT include an intercept by default)
X = sm.add_constant(X)

# 4. Fit the Ordinary Least Squares model
model = sm.OLS(y, X).fit()

# 5. View regression results
print(model.summary())
```
